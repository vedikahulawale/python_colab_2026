Q3. Count vowels in a string




Write a function count_vowels(text) that counts vowels (a, e, i, o, u) in a string. The function should ignore uppercase and lowercase differences. Do not use the built-in count() method.

```python
def count_vowels(t):
  c=0
  for i in t:
    if i.lower() in ('a','e','i','o','u'):
      c=c+1
  return c

count_vowels("SrUshti")

```

Q4. Find the sum of digits




Write a function sum_digits(n) that returns the sum of all digits in a non-negative integer. Solve it using logic rather than converting the number into a string.

```python
def sum_digits(*n):
  s=0
  for i in n:
    for j in i:
      if j>0:
        s=s+j
    return s
n=(1,2,3)
sum_digits(n)


```

Q7. Find students who passed using **kwargs




Write a function passed_students(**scores) that accepts student names and marks as keyword arguments. Return a dictionary containing only students who scored 40 or more.

Q8. Generate prime numbers in a range




Write a function generate_primes(start, end) that returns all prime numbers between start and end, including both endpoints when applicable.

```python
def generate_primes(s=1,e=1):
  l=[]
  for i in range(s,e+1):
    f=1
    for j in range(2,i):
      if i%j==0:
        f=0
        break

    if f==1 and i>1:
      l.append(i)
  return l
generate_primes(2,27)
```

Q9. Calculate factorial using recursion





Create a recursive function factorial(n) that calculates the factorial of a non-negative integer without using math.factorial().

```python
def factorial(n):
  f=1
  for i in range(1,n+1):
    f=f*i
  return f
factorial(6)


```

Q10. Check a palindrome using recursion <br>
Write a function is_palindrome(text) that checks whether a string is a palindrome using recursion. Ignore spaces and letter case.

```python
def is_palindrome(t):
  r=''
  p=''
  for i in t:
    r=r.lower()
    if i!=" ":
      r=r+i
  for i in r:
    p=i+p
  if r==p:
    return "It's palindrome text"
  else:
    return "It's not palindrome text"
is_palindrome("MaDam  ")
```

Q12. Filter words by their length <br>
Write a function filter_long_words(words, min_length) that uses filter() and a lambda function to return only words whose lengths are greater than or equal to min_length.

```python
def filter_long_words(w,ml=1):

  return list(filter(lambda w:len(w)>=ml,w))

w=["apple", "cat", "banana", "dog", "elephant"]
filter_long_words(w,5)
```

Q13. Multiply all numbers using reduce() <br>
Write a function product_of_numbers(numbers) that uses functools.reduce() to calculate the product of all numbers in a list.

```python
from functools import reduce
def product_of_numbers(*n):
   r=reduce(lambda x,y:x*y,n)
   return r
product_of_numbers(1,2,3)
```
