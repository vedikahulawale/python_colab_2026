Q2. Check even or odd




Create a function is_even(n) that returns True if a number is even; otherwise, return False.

```python
def is_even(n):
  if n%2==0:
    return True
  else:
    return False
is_even(5)
```

Q3. Count vowels in a string




Write a function count_vowels(text) that counts vowels (a, e, i, o, u) in a string. The function should ignore uppercase and lowercase differences. Do not use the built-in count() method.

```python
def count_vowels(t):
  c=0
  for i in t:
    if i.lower() in ('a','e','i','o','u'):
      c=c+1
  return c

count_vowels("SrUshti")

```

Q4. Find the sum of digits




Write a function sum_digits(n) that returns the sum of all digits in a non-negative integer. Solve it using logic rather than converting the number into a string.

```python
def sum_digits(*n):
  s=0
  for i in n:
    for j in i:
      if j>0:
        s=s+j
    return s
n=(1,2,3)
sum_digits(n)


```

Q6. Sum any number of arguments
Create a function sum_all(*args) that returns the sum of any number of numeric arguments. It should also work when no arguments are passed.

```python

def sum_all(*a):
  s=0
  for i in a:
    s=s+i
  return s

sum_all(1,2)


```

Q7. Find students who passed using **kwargs




Write a function passed_students(**scores) that accepts student names and marks as keyword arguments. Return a dictionary containing only students who scored 40 or more.

```python
def passed_students(**s):
  d={}
  for i,j in s.items():
    if j>=40:
      d[i]=j
  return d
passed_students(vedika=100,sai=80,xyz=76,ey=78,th=23)
```

Q8. Generate prime numbers in a range




Write a function generate_primes(start, end) that returns all prime numbers between start and end, including both endpoints when applicable.

```python
def generate_primes(s=1,e=1):
  for i in range(s,e+1):
    f=1
    l=[]
    for j in range(2,i):
      if i%j==0:
        f=0
        break
    if f==1:
      l.apppend(i)
  return l

```
