## Q1. Student Marks

**Level:** Beginner  
**Concept:** Traditional getter and setter methods

Create a `Student` class with these requirements:

1. The constructor initializes a student's name and marks.
2. Store marks in an attribute named `_marks`.
3. Create `get_marks()` to return the marks.
4. Create `set_marks()` to update the marks.
5. Marks must be between 0 and 100, inclusive. Reject invalid values without changing the existing marks.

**Test:** Initial marks = `75`; update to `90`; try setting `120`; check the marks after the invalid update.

**Critical thinking:** Why should the original value remain unchanged when a new value is invalid?

```python
# Write your solution here
class Student:
  def __init__(self,stu_name:str,marks:int):
    self.stu_name=stu_name
    self._marks=marks

  def get_marks(self):
    return self._marks

  def set_marks(self,marks:int):
    if marks<=100 and marks>=0:
      self._marks=marks
    else:
      print("Enter the correct marks")







s=Student('vedika',50)
s.set_marks(0)
s.get_marks()

```

## Q2. Student Marks Using `@property`

**Level:** Beginner  
**Concept:** Pythonic getter and setter

Rewrite Q1 using the `@property` decorator instead of `get_marks()` and `set_marks()` methods.

1. Store the value in `_marks`.
2. Use `@property` to retrieve marks.
3. Use `@marks.setter` to update marks.
4. Keep the same validation rule.
5. Read and update the value using `student.marks`.

**Critical thinking:** What is the advantage of `student.marks` over explicitly calling getter and setter methods?

```python
# Write your solution here
class Student:
  def __init__(self,marks):
    self._marks=marks

  @property
  def marks(self):
    return self._marks

  @marks.setter
  def marks(self,new_marks):
    self._marks=new_marks

s1=Student(40)
print(s1.marks)

s1.marks = 95

print(s1.marks)
```

## Q3. Employee Salary Validation

**Level:** Intermediate  
**Concept:** Encapsulation, validation, computed and read-only properties

Create an `Employee` class with these requirements:

1. Store `name` and `salary`.
2. Use `@property` and a setter for salary.
3. Salary must be greater than zero.
4. Reject negative salary, zero, and invalid data types such as a string.
5. Add a read-only `annual_salary` property that calculates the salary for 12 months.

**Test:** Salary = `50000`; change to `60000`; then try `-1000`.

**Critical thinking:** If monthly salary changes, should annual salary update automatically? How can you avoid storing a second, potentially outdated value?

```python
# Write your solution here
class Employee:
  def __init__(self,name,salary):
    self.name=name
    self.salary=salary


  @property
  def salary(self):
    return self.salary

  @salary.setter
  def salary(self,salary:float):
    if salary>0 and (type(salary)==float or type(salary)==int):
      self.salary=salary

  def annual_salary(self):
    anns=self.salary*12
    return anns

e=Employee("vedika",2)





```
