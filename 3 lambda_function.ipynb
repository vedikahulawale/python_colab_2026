Q1. Square a number <br>
Write a lambda function that takes a number and returns its square.

```python
sq=lambda x:x**2
sq(3)
```

Q2. Add two numbers <br>
Create a lambda function that accepts two numbers and returns their sum.

```python
s=lambda x,y:x+y
s(2,3)
```

Q3. Check even or odd <br>
Write a lambda function that returns "Even" if a number is even and "Odd" otherwise.

```python
is_even_odd=lambda x:"even" if x%2==0 else "odd"
is_even_odd(4)
```

Q4. Find the greater number <br>
Create a lambda function that accepts two numbers and returns the larger one. If both are equal, return either number.

```python
max_n=lambda x,y:y if x<y else x
max_n(44,44)
```

Q5. Convert Celsius to Fahrenheit <br>
Write a lambda function to convert Celsius into Fahrenheit using:

Q6. Positive, negative, or zero <br>
Create a lambda function that returns "Positive", "Negative", or "Zero" depending on the input number.

```python
check= lambda n: 'positive' if n>0 else ('negative'if n<0 else 'zero')
check(0)
```

Q7. Calculate a discount <br>
A shop gives a 10% discount when the purchase amount is at least ₹1,000. Write a lambda function that returns the final amount after the discount, or the original amount otherwise.

```python
shop=lambda a:a-(a*0.1) if a>=1000 else a
shop(2000)
```

Q8. Square every number <br>
Given a list of numbers, use map() and a lambda function to return a list containing the square of every number.

```python

l=[1,2,3,4]
s=map(lambda x:x**2,l)
list(s)
```

Q9. Filter even numbers <br>
