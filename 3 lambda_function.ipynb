Q7. Calculate a discount <br>
A shop gives a 10% discount when the purchase amount is at least ₹1,000. Write a lambda function that returns the final amount after the discount, or the original amount otherwise.

```python
shop=lambda a:a-(a*0.1) if a>=1000 else a
shop(2000)
```

Q14. Keep names starting with a vowel <br>
Use filter() and lambda to return only the names that begin with a vowel. Treat uppercase and lowercase vowels equally.

```python
l=['abc','sc','avet']
v=["a",'e','i','o',"u"]
r=list(filter(lambda n:n[0].lower() in v,l))
r
```

Q15. Sort names alphabetically by their last character <br>
Sort a list of names according to the last character of each name. If two names have the same last character, preserve a sensible ordering by using the full name as a secondary key.

```python
l=["Ravi", "Mona", "Amit", "Neha"]
r=sorted(l,key=lambda n: (n[-1],n))
r
```
