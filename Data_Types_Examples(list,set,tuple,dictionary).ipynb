## Q39 Find repeated words in a sentence using a set.

---



```python
s="vedika hulawale jftt njhj mnb vedika hulawale njhj"
r=''
rw=set()
w=set()
for i in s:
  if i!=" ":
    r=r+i
  else:
    if r in w:
      rw.add(r)
    else:
      w.add(r)
    r=''
if r != "":
    if r in w:
      rw.add(r)
    else:
      w.add(r)
rw
```
