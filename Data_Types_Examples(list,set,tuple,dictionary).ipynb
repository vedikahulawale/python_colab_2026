## Q16 Create a set of all vowels present in a string.

---



```python
s="vedika"
v=set()
for i in s:
  if i.lower() in ['a','e','i','o','u']:
    v.add(i)
v
```

## Q18 Find unique lowercase characters from a mixed-case string.

```python
s="Apple Pie"
c=set()
for i in s:
  if i.islower():
    c.add(i)
c
```
