## Q4 Check whether 10 exists in a set.

```python
s={1,2,3,4,5,3,1,10}
for i in s:
  if i==10:
    print(f"10 is present in set")
```

## Q15 Create a set containing squares from 1 to 10.

---



```python
s=set()
for i in range(1,11):
  s.add(i**2)
s
```

## Q16 Create a set of all vowels present in a string.

---



```python
s="vedika"
v=set()
for i in s:
  if i in ['a','e','i','o','u']:
    v.add(i)
v
```
