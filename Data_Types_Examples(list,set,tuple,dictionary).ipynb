## Q1 Create a set containing the numbers 1,2,3,4,5.

---



```python
s={1,2,3,4,5}
print(s)
```

## Q2 Convert a list [1,2,2,3,3,4] into a set.

---



```python
s=set([1,2,2,3,3,4])
s
```

## Q3 Find the number of unique elements in a list using a set.

---



```python
l=[1,2,2,3,3,4]
s=set(l)
s
```

## Q4 Check whether 10 exists in a set.

---



```python
s={1,2,3,4,5,3,1,10}
for i in s:
  if i==10:
    print(f"10 is present in set")
```

## Q5 Add an element to a set.

---



```python
s={1,2,3,4}
s.add(5)
s
```

## Q6 Remove a given element from a set using remove().

---



```python
s={1,2,3,4}
s.remove(4) # not present element it raise the error
s
```

## Q7 Remove an element safely using discard().

---



```python
s={1,2,3,4}
s.discard(4) # not present element it not raise the error
s
```

## Q8 Find the length of a set.

---



```python
s={1,2,3,4,4,3,3}
l=len(s)
l
```

## Q9 Iterate through all elements of a set.

---



```python
s={1,2,3,4}
for i in s:
  print(i)
```
